```python
import numpy as np
units = np.array(["Punjab","Sindh","KPK","Balochistan",
                  "Islamabad","AJK","Gilgit-Baltistan"])
pop_2023=np.array([127688922,55696147,40856097,
                   14894402,2363863,4333467,1709049])
pop_2017=np.array([110012442,47886051,30508737,
                   12344408,2001579,4069732,1493000])
print("a) Total Population:",pop_2023.sum())
print("b) Population Share (%):")
print(pop_2023 / pop_2023.sum()*100)
print("c) Most Populated Unit:",units[pop_2023.argmax()])
print("d) Above 20 Million:")
print(units[pop_2023 > 20000000])
growth = ((pop_2023 - pop_2017) / pop_2017) / 6 * 100
print("e) Annual Growth (%):",growth)
print("f) Fastest Growing Unit:",units[growth.argmax()])

      
```

```python
import numpy as np

tmax = np.array([
    [25, 28, 32, 35],
    [22, 26, 30, 33],
    [20, 25, 31, 36]
])

rain = np.array([
    [10, 20, 30, 40],
    [15, 25, 35, 45],
    [12, 22, 32, 42]
])

print("a) Shape:", tmax.shape)
print("b) Average:", np.mean(tmax, axis=1))
print("c) Maximum:", np.max(tmax, axis=1))
print("d) Monthly Average:", np.mean(tmax, axis=0))
print("e) Above 30:", tmax[tmax > 30])
print("f) Fahrenheit:", tmax[0] * 9/5 + 32)
print("g) Rain:", rain[:, 1:4])
print("h) Total Rain:", np.sum(rain, axis=1))
```

```python
import numpy as np

cpi = np.array([
    [28.3, 27.4, 26.9, 28.3, 31.4, 29.7, 28.3, 27.4, 26.0, 17.3],
    [11.1, 9.6, 7.2, 4.9, 4.1, 2.4, 0.3, 0.7, 0.3, 0.3]
])

cpi_months = np.array([
    "Jul", "Aug", "Sep", "Oct", "Nov",
    "Dec", "Jan", "Feb", "Mar", "Apr"
])

print("a) Average:", np.mean(cpi, axis=1))
print("b) Above 25%:", cpi_months[cpi[0] > 25])

drop = cpi[0] - cpi[1]
print("c) Monthly Drop:", drop)

print("d) Biggest Drop:", cpi_months[np.argmax(drop)])

labels = np.where(cpi[1] < 5, "Low",
         np.where(cpi[1] < 10, "Moderate", "High"))

print("e) Labels:", labels)
```

```python
import numpy as np

marks = np.array([45, 28, 60, 32, 75, 40, 29, 90])

new_marks = np.clip(marks + 5, 0, 100)

print("T9a New Marks:", new_marks)

pass_percent = (new_marks >= 33).mean() * 100
print("T9b Pass Percentage:", pass_percent)

grace_pass = marks[(marks < 33) & (new_marks >= 33)]
print("T9c Grace Pass:", grace_pass)

z = (marks - marks.mean()) / marks.std()
print("T9d Z-Scores:", z)
```

```python
import numpy as np

image = np.array([
    [255, 100, 50, 200],
    [150, 255, 0, 100],
    [50, 100, 255, 150],
    [200, 50, 100, 255]
])

print("Original Image:")
print(image)

print("Image Shape:", image.shape)
print("Total Pixels:", image.size)
print("Maximum Pixel Value:", image.max())
print("Minimum Pixel Value:", image.min())
print("Average Pixel Value:", image.mean())

bright_image = image + 10
print("Bright Image:")
print(bright_image)
```
